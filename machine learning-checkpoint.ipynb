## 2. Exploratory Data Analysis

```python
# Display summary statistics for numerical columns
df.describe()
```

```python
import matplotlib.pyplot as plt

# Plot the distribution of house prices
plt.figure(figsize=(8, 5))
plt.hist(df['price'], bins=30, edgecolor='black')

plt.title('Distribution of House Prices')
plt.xlabel('House Price ($)')
plt.ylabel('Frequency')
plt.show()
```

**Observation – House Price Distribution**

The histogram shows that most houses are in the lower price range, while only a few have very high prices. This makes the distribution right-skewed. The average price is around $540,089, which is higher than the median of $450,000. Some houses cost as much as $7.7 million, so these expensive properties may affect the prediction results.

```python
# Examine the relationship between living area and house price
plt.figure(figsize=(8, 5))
plt.scatter(df['sqft_living'], df['price'], alpha=0.3)

plt.title('Living Area vs House Price')
plt.xlabel('Living Area (sq ft)')
plt.ylabel('House Price ($)')
plt.show()
```

**Observation – Living Area vs House Price**

The scatter plot shows that bigger houses generally have higher prices. However, some houses with similar living areas have different prices, so house size is not the only factor affecting the price. There are also a few properties that stand out from the rest. This suggests that living area could be a useful feature when predicting house prices.
